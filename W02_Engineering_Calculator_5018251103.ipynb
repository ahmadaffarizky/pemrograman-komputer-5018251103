# Short Reflection


1. Apa perbedaan int dan float?

int digunakan untuk menyimpan bilangan bulat, seperti 5, 10, atau -3. Sedangkan float digunakan untuk menyimpan bilangan yang memiliki nilai desimal, seperti 5.5, 10.25, atau -3.2.

2. Mengapa hasil input() perlu dikonversi menggunakan float() untuk perhitungan numerik?

Karena hasil dari input() secara default berupa string (teks). Agar dapat digunakan dalam perhitungan matematika, data tersebut harus diubah menjadi tipe numerik seperti float(). Contohnya, float(input()) mengubah input menjadi bilangan desimal yang dapat dihitung.

3. Apa perbedaan operator = dan == dalam Python?


= digunakan untuk memberikan atau menyimpan nilai ke dalam variabel. Sedangkan == digunakan untuk membandingkan apakah dua nilai sama.

4. Apa fungsi operator and dan or?

and digunakan ketika semua kondisi harus bernilai True agar hasilnya True. Sedangkan or digunakan ketika minimal salah satu kondisi bernilai True agar hasilnya True.

5. Apa satu kesalahan yang Anda temui saat mengerjakan tugas ini, dan bagaimana Anda memperbaikinya?


Salah satu kesalahan yang saya temui adalah lupa mengonversi hasil input() menjadi float, sehingga data input masih dianggap sebagai teks dan tidak dapat digunakan untuk perhitungan. Saya memperbaikinya dengan menggunakan float(input()) pada setiap input numerik.
